```python
import pandas as pd
import random

from datetime import datetime, timedelta
```

```python

random.seed(10)

nombres = ["Carlos", "Manuel", "Saul", "Lucía", "María", "Pedro", "Ana", "Jorge", "Sofía", "Luis"]
apellidos = ["Ramirez", "Fernandes", "Sosa", "García", "Torres", "Lopez", "Martinez", "Rojas", "Castillo", "Vega"]

# Generar 200 DNIs únicos
dnis_unicos = random.sample(range(10000000, 99999999), 200)

data = []
for i in range(1, 201):
    nombre = random.choice(nombres)
    apellido = random.choice(apellidos)
    dni = str(dnis_unicos[i-1])  # siempre único
    telefono = "9" + str(random.randint(10000000, 99999999))
    direccion = "Calle " + str(random.randint(1, 200)) + " #" + str(random.randint(100, 999))

    data.append([i, nombre, apellido, dni, telefono, direccion])

df = pd.DataFrame(data, columns=["IdCliente","Nombre","Apellido","DNI","Telefono","Direccion"])

print(df.head())
```

```python
df.to_csv("ClienteServicioTecnico.csv",index=False)
```

```python
random.seed(30)

num_clientes = 200
num_tecnicos = 50
num_equipos = 100

estados = ["Pendiente", "En proceso", "Finalizado", "Cancelado"]
observaciones = [
    "Revisión inicial del equipo",
    "Cambio de pieza principal",
    "Actualización de software",
    "Limpieza y mantenimiento",
    "Diagnóstico completo",
    "Cliente solicita revisión adicional",
    "Equipo entregado sin fallas",
    "Pendiente de repuesto"
]

data = []
for i in range(1, 101):
    id_cliente = random.randint(1, num_clientes)
    id_tecnico = random.randint(1, num_tecnicos)
    id_equipo = random.randint(1, num_equipos)
    fecha = datetime(2026, 7, 1) + timedelta(days=random.randint(0, 30))
    estado = random.choice(estados)
    observacion = random.choice(observaciones)

    data.append([
        id_cliente,
        id_tecnico,
        id_equipo,
        fecha.strftime("%Y-%m-%d"),
        estado,
        observacion
    ])

# Crear DataFrame (sin IdOrden porque es IDENTITY)
df = pd.DataFrame(data, columns=["IdCliente","IdTecnico","IdEquipo","FechaIngreso","Estado","Observacion"])
print(df.head())
```

```python
import pandas as pd
import random

# Cargar tu archivo original
df = pd.read_csv("ClienteServicioTecnico.csv")

# Detectar DNIs repetidos
conteo = df["DNI"].value_counts()
repetidos = conteo[conteo > 1].index

# Generar nuevos DNIs únicos para reemplazar los repetidos
usados = set(df["DNI"])
for idx in df[df["DNI"].isin(repetidos)].index:
    while True:
        nuevo_dni = str(random.randint(10000000, 99999999))
        if nuevo_dni not in usados:
            df.at[idx, "DNI"] = nuevo_dni
            usados.add(nuevo_dni)
            break

# Reasignar IdCliente consecutivo
df["IdCliente"] = range(1, len(df) + 1)
```

```python
import pandas as pd
import random

# Cargar tu archivo
df = pd.read_csv("ClienteServicioTecnico.csv")

# Guardar los DNIs ya usados
usados = set(df["DNI"])

# Detectar filas con DNIs repetidos
duplicados = df[df.duplicated(subset="DNI", keep=False)]

# Reemplazar los DNIs repetidos con nuevos únicos
for idx in duplicados.index:
    while True:
        nuevo_dni = str(random.randint(10000000, 99999999))
        if nuevo_dni not in usados:
            df.at[idx, "DNI"] = nuevo_dni
            usados.add(nuevo_dni)
            break

# Reasignar IdCliente consecutivo
df["IdCliente"] = range(1, len(df) + 1)

# Guardar archivo corregido
df.to_csv("ClienteCorregido.csv", index=False)
```

```python
df = pd.read_csv("ClienteFinal.csv")

# Crear un conjunto con los DNIs ya usados
usados = set()

# Recorrer todas las filas y corregir si hay repetidos
for idx in df.index:
    dni = str(df.at[idx, "DNI"])
    if dni in usados:
        # Generar un nuevo DNI único
        while True:
            nuevo_dni = str(random.randint(10000000, 99999999))
            if nuevo_dni not in usados:
                df.at[idx, "DNI"] = nuevo_dni
                usados.add(nuevo_dni)
                break
    else:
        usados.add(dni)

# Reasignar IdCliente consecutivo
df["IdCliente"] = range(1, len(df) + 1)

# Guardar archivo limpio
df.to_csv("ClienteUnico.csv", index=False)
```
